
# Creating a Realtime Currency Converter

```python
import os
from langchain_groq import ChatGroq
from langchain_core.tools import tool, InjectedToolArg
from typing import Annotated
import requests
from langchain_core.messages import HumanMessage
import json
```

```python
load_dotenv()

if not os.getenv("GROQ_API_KEY"):
  raise ValueError("Set GROQ_API_KEY in your local .env file")

llm = ChatGroq(model='llama-3.3-70b-versatile')
```

```python
# creating tools

EXCHANGE_RATE_API_KEY = os.getenv("EXCHANGE_RATE_API_KEY")
if not EXCHANGE_RATE_API_KEY:
  raise ValueError("Set EXCHANGE_RATE_API_KEY in your local .env file")
@tool
def get_conversion_fector(base_currency: str, target_currency: str) -> float:
  """ This function fetches the currency conversion factor between a given base currency and a target currency.
  """
  url = f"https://v6.exchangerate-api.com/v6/{EXCHANGE_RATE_API_KEY}/pair/{base_currency}/{target_currency}"
  response = requests.get(url)

  return response.json()

@tool
def convert(base_currency_value: int|float , conversion_rate: Annotated[float, InjectedToolArg]) -> float:
  """ Giveen a currency convesion rate this function calculate the target corrency value from a given base currency value"""
  return base_currency_value * conversion_rate
```

```python
convert.args
```

```python
get_conversion_fector.invoke({'base_currency':'INR', 'target_currency': 'USD'})
```

```python
llm_with_tools = llm.bind_tools([get_conversion_fector, convert])
```

```python
message = [HumanMessage('What is the conversion factor between india and usa, and based on that can you convert 10 ind to usr')]
```

```python
ai_msg = llm_with_tools.invoke(message)
message.append(ai_msg)
```

```python
ai_msg.tool_calls
```

```python
# logic of conversion

for tool_call in ai_msg.tool_calls:
  # execute the first tool and get the value of conversion rate
  if tool_call['name'] == 'get_conversion_fector':
    tool_message1 = get_conversion_fector.invoke(tool_call)
    conversion_rate = json.loads(tool_message1.content)['conversion_rate']
    message.append(tool_message1)
  # execute the second tool using conversion rate from tool
  if tool_call['name'] == 'convert':
    # fetch the current arg
    tool_call['args']['conversion_rate'] = conversion_rate
    tool_message2 = convert.invoke(tool_call)
    message.append(tool_message2)
```

```python
message
```

```python
result = llm_with_tools.invoke(message)
```
